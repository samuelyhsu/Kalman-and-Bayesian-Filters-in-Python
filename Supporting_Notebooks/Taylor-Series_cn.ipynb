# 用泰勒级数线性化

泰勒级数把函数表示为无穷多项之和。即使对于非线性函数，这些项也是线性的，所以我们可以用线性代数来表示任意非线性函数。这种选择的代价是：除非使用无穷多项，否则我们计算出的值是近似值而不是精确值。

实函数或复函数 f(x) 在 x=a 处的泰勒级数是无穷级数

$$f(x) = f(a) + f'(a)(x-a) + \frac{f''(a)}{2!}(x-a)^2 + \, ...\,  + \frac{f^{(n)}(a)}{n!}(x-a)^n + \, ...$$

其中 $f^{n}$ 是 f 的 n 阶导数。要计算 $f(x)=sin(x)$ 在 $x=0$ 处的泰勒级数，我们先求出 f 的各项。

$$\begin{aligned}
f^{0}(x) &= sin(x) ,\ \  &f^{0}(0) &= 0 \\
f^{1}(x) &= cos(x),\ \  &f^{1}(0) &= 1 \\
f^{2}(x) &= -sin(x),\ \  &f^{2}(0) &= 0 \\
f^{3}(x) &= -cos(x),\ \  &f^{3}(0) &= -1 \\
f^{4}(x) &= sin(x),\ \  &f^{4}(0) &= 0 \\
f^{5}(x) &= cos(x),\ \  &f^{5}(0) &= 1
\end{aligned}
$$

现在可以把这些值代入公式。

$$\sin(x) = \frac{0}{0!}(x)^0 + \frac{1}{1!}(x)^1 + \frac{0}{2!}(x)^2 + \frac{-1}{3!}(x)^3 + \frac{0}{4!}(x)^4 + \frac{1}{5!}(x)^5 + ... $$

再用一些代码来测试：

```python
import numpy as np

x = .3
estimate = x - x**3/6 + x**5/120
exact = np.sin(.3)

print('estimate of sin(.3) is', estimate)
print('exact value of sin(.3) is', exact)
```

对于只有三项来说，这已经不错了。如果你感兴趣，不妨把它实现为一个 Python 函数，用来计算任意项数的级数。
