# 安装

本书是用 Jupyter Notebook 写成的，这是一个基于浏览器的交互式 Python 环境，可以把 Python、文字和数学混合在一起。我选择它是因为它的交互特性——在开始使用交互式环境之前，我觉得卡尔曼滤波几乎不可能学会。除非你能修改许多参数并立即看到输出，否则很难对它们形成直觉。交互式环境还允许你尝试“如果……会怎样”的场景。“如果我把 $\mathbf{Q}$ 设为零会怎样？”用 Jupyter Notebook 很容易找到答案。

我选择它的另一个原因是，大多数教科书会让许多事情变得不透明。例如，伪代码旁边可能有一张漂亮的图。这张图是由软件生成的，但读者却得不到这个软件。我希望制作本书所用的一切都能提供给你。如何绘制协方差椭圆？读大多数书你都不会知道。而使用 Jupyter Notebook，你只需要看一下源代码。

即使你选择在线阅读本书，你也需要安装 Python 和 SciPy 技术栈，以便编写自己的卡尔曼滤波器。安装这些库有许多不同的方式，我无法一一涵盖，但我会介绍几种典型场景。

## SymPy

SymPy 是一个用于执行符号数学运算的 Python 包。它的全部功能超出了本书的范围，但它可以进行代数运算、对方程求积分和求导、求微分方程的解，等等。例如，我们用它来计算矩阵的雅可比矩阵，以及期望值的积分计算。

首先来看一个简单的例子。我们将导入 SymPy，初始化它的美化打印功能（它会用 LaTeX 打印方程）。然后我们将声明一个供 SymPy 使用的符号。

```python
import sympy
sympy.init_printing(use_latex='mathjax')

phi, x = sympy.symbols('\phi, x')
phi
```

注意它是如何用 LaTeX 打印符号 `phi` 的。现在我们来做一些数学运算。$\sqrt{\phi}$ 的导数是什么？

```python
sympy.diff('sqrt(phi)')
```

我们可以对方程进行因式分解

```python
sympy.factor(phi**3 -phi**2 + phi - 1)
```

也可以把它们展开。

```python
((phi+1)*(phi-4)).expand()
```

你可以为方程中的变量代入具体的值来求值：

```python
w =x**2 -3*x +4
print(w.subs(x, 4))
print(w.subs(x, 12))
```

对于使用了尚未定义的符号的方程，你也可以使用字符串：

```python
x = sympy.expand('(t+1)*2')
x
```

现在我们用 SymPy 来计算一个矩阵的雅可比矩阵。给定函数

$$h=\sqrt{(x^2 + z^2)}$$

求它关于 x、y 和 z 的雅可比矩阵。

```python
x, y, z = sympy.symbols('x y z')

H = sympy.Matrix([sympy.sqrt(x**2 + z**2)])

state = sympy.Matrix([x, y, z])
H.jacobian(state)
```

现在，给定连续过程噪声矩阵，我们来计算离散过程噪声矩阵 $\mathbf Q$
$$\mathbf Q = \Phi_s \begin{bmatrix}0&0&0\\0&0&0\\0&0&1\end{bmatrix}$$

其积分为

$$\mathbf Q = \int_0^{\Delta t} \mathbf F(t)\mathbf Q\mathbf F^T(t)\, dt$$

其中
$$\mathbf F(\Delta t) = \begin{bmatrix}1 & \Delta t & {\Delta t}^2/2 \\ 0 & 1 & \Delta t\\ 0& 0& 1\end{bmatrix}$$

```python
dt = sympy.symbols('\Delta{t}')
F_k = sympy.Matrix([[1, dt, dt**2/2],
                    [0,  1,      dt],
                    [0,  0,       1]])
Q = sympy.Matrix([[0,0,0],
                  [0,0,0],
                  [0,0,1]])

sympy.integrate(F_k*Q*F_k.T,(dt, 0, dt))
```
