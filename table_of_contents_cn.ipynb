[**第 6 章：多元卡尔曼滤波器**](./06-Multivariate-Kalman-Filters_cn.ipynb)

我们把一元章节中建立的卡尔曼滤波器扩展为适用于线性问题的完整通用滤波器。读完这一章，你将理解卡尔曼滤波器如何工作，以及如何为你选择的（线性）问题设计并实现一个卡尔曼滤波器。

[**第 9 章：非线性滤波**](./09-Nonlinear-Filtering_cn.ipynb)

前面介绍的卡尔曼滤波器只适用于线性问题，然而世界是非线性的。这里我介绍非线性系统给滤波器带来的问题，并简要讨论我们将在后续章节中学习的各种算法。


[**第 10 章：无迹卡尔曼滤波器**](./10-Unscented-Kalman-Filter_cn.ipynb)

无迹卡尔曼滤波器（UKF）是卡尔曼滤波理论中较新的发展。它使你能够对非线性问题进行滤波，而不需要像扩展卡尔曼滤波器那样要求有闭式解。

现有教材通常要么不提这个主题，要么一笔带过，而把大部分篇幅用来讨论扩展卡尔曼滤波器。我把它放在前面，是因为 UKF 理解和实现起来都简单得多，而且滤波性能通常与扩展卡尔曼滤波器相当甚至更好。对于实际问题，我总是先尝试实现 UKF，你也应该这样做。


[**第 11 章：扩展卡尔曼滤波器**](./11-Extended-Kalman-Filters_cn.ipynb)

扩展卡尔曼滤波器（EKF）是把非线性问题线性化最常用的方法。现实世界中的大多数卡尔曼滤波器都是 EKF，所以你需要理解这些内容，才能看懂现有的代码、论文、演讲等。


[**第 12 章：粒子滤波器**](./12-Particle-Filters_cn.ipynb)

粒子滤波器使用蒙特卡洛技术对数据进行滤波。它们能轻松处理高度非线性和非高斯的系统，以及多峰分布（同时跟踪多个目标），代价是计算需求很高。


[**第 13 章：平滑**](./13-Smoothing_cn.ipynb)

卡尔曼滤波器是递归的，因此非常适合实时滤波。然而，它们在对数据进行事后处理时同样表现极佳。毕竟，卡尔曼滤波器是“预测-校正”式的，而预测过去比预测未来更容易！我们讨论一些常用的方法。


[**第 14 章：自适应滤波**](./14-Adaptive-Filtering_cn.ipynb)

卡尔曼滤波器假定只有单一的过程模型，但机动目标通常需要用几种不同的过程模型来描述。自适应滤波使用多种技术，使卡尔曼滤波器能够适应目标不断变化的行为。
